```python
import pandas as pd
import numpy as np

data = {
    "name" :  ["ananya rao", "RAHUL Verma", "Priya Nair", "Priya Nair", "sana khan", "Vikaram  Singh", "Karan Mehta", "meera IYER", "Arjun Das", None],
    "city" : ["Bengaluru", "mumbai", "Chennai", "Chennai", "PUNE", "bengaluru", "Mumbai", "chennai", "Pune", "Bengaluru"],
    "math_score" : [78, 55, 92, 92, np.nan ,64, 71, 95, np.nan, 49]
}

raw_students = pd.DataFrame(data)
print(raw_students)
```

```python
#DataFrame.isnull
print(raw_students.isnull())
```

```python
#dataframe.isnull().sum()
print(raw_students.isnull().sum())
```

```python
print(raw_students["math_score"].isnull())
```

```python
print(raw_students["math_score"].isnull().sum())
```

```python
#write a signle line that check whether the city column has any mission value at all
X = raw_students["city"].isnull().sum()
if X == 0 :
  print("No null values found")
else :
  print(X, "null values found")
```

```python
#write a signle line that check whether the city column has any mission value at all - just true or false,  not count
print(raw_students["city"].isnull().any())
```

```python
print(raw_students["name"].isnull().any())
```

```python
# to get the count of total null values in the dataframe we use .sum().sum()
print(raw_students.isnull().sum().sum())
```

```python
# option 1: drop rows with any missing value

dropped = raw_students.dropna()
print(dropped)

print("Rows before" , len(raw_students))
print("Rows after" , len(dropped))
```

```python
drop_column = raw_students.dropna(axis=1) # axis = 1 drop column and axis = 0 drop row
print(drop_column)
```

```python
# option 2: fill the missing math_score with the column average instead of dropping

average_score = raw_students["math_score"].mean()
print(average_score)

filled  = raw_students.copy()
# print(filled)

#DataFrame["column"] = dataframe["column"].fillna(value)

filled["math_score"] = filled["math_score"].fillna(average_score)
print(filled)
```

```python
print(raw_students.duplicated())
```

```python
#dataframe.drop_duplicates() - returns a new dataframe with every flagged duplicate removed, keeping only the first occurence of each repeated row
cleaned = raw_students.drop_duplicates()
print(cleaned)

print("Rows before :", len(raw_students))
print("Rows after :", len(cleaned))
```

```python
#checking duplicated on just one column, instead of the whole row
# dataframe.duplicated(subset = ["column_name"])

print(raw_students.duplicated(subset = ["name", "math_score"]))
```

```python
# .unique() - return every distinct value that actually appears in the provided columns with no repeats
# -- the direct way to SEE a messay data before cleaning it
print(raw_students["city"].unique())
```

```python
# .str.strip().str.title()
# dataframe["column_name"].str.strip().str.title()

# .str -> tells the pandas to apply the method that follows to every single value in the column, one at a time
# .strip() -> the exact same string method from the earlier sessions -remove leading/ trailing whitespaces
# .title() -> also familiar -capatilises each word consistetly

raw_students["city_clean"] = raw_students["city"].str.strip().str.title()
print(raw_students["city_clean"])
print(raw_students["city_clean"].unique())
```

```python
# one more tool : explicitly converting a column's type

# syntax : -- .astype(value)
# .astype(float) -- explicitly converts every value in the column to the named type
print(raw_students.dtypes)
```

```python
filled["math_score"] = filled["math_score"].astype(int)
print(filled.dtypes)
```

```python
# standardise all the city name in lower case and give me the unique values
print(raw_students["city"].str.strip().str.lower().unique())    # .lower for lowercase
```

```python
print(raw_students["name"].tolist())
```

```python
# a full .str cleaning chain
# dataframe["column"].fillna("unknown").str.strip().str.split().str.join(" ").str.title()
# fillna("unknown") - comes First - handles any missing value before any string method tries to run on it and fails
# .str.strip() - removes the leading/trailing whitespaces
# .str.split() - breaks each name into a list of words, automatically collapsing any extra internal spaces in the process
# .str.join() - it sticks back the words together with exaclty one space between them
# .str.title() - finishes the consistent capatilisation (title case)

raw_students["name_clean"] = (
    raw_students["name"].
    fillna("Unknown")
    .str.strip()
    .str.split()
    .str.join(" ")
    .str.title()
)
print(raw_students["name_clean"])
```
