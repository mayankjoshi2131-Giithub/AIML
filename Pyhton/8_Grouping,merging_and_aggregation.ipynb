```python
import pandas as pd

orders = pd.DataFrame({
    "order_id" : [101, 102, 103, 104, 105, 106, 107, 108],
    "customer_id" : [1, 2, 1, 3, 2, 1, 4, 3],
    "products" : ["Notebook", "Pen Set", "Desk Lamp", "Backpack", "Mouse", "Notebook", "Chair", "Pen Set"],
    "amount" : [45, 120, 899, 1299, 350, 45, 560, 120],
    "city" : ["Bengaluru", "Mumbai", "Bengaluru",  "Chennai", "Mumbai", "Bengaluru", "Pune", "Chennai"]

})

customers = pd.DataFrame({
    "cust_id" : [1, 2, 3, 5],
    "customer_name" : ["Ananya", "Rahul", "Priya", "Vikram"],
    "segment" : ["Retail", "Retail","Wholesale", "Ratail"]
})

print(orders)
print(customers)
```

```python
# findout the order amount per city
print(orders.groupby("city")["amount"].sum())
```

```python
# group by products instead of city and find the total amount solder per product
print(orders.groupby("products")["amount"].sum())
```

```python
# write a line that show the maximum single order amount per city
print(orders.groupby("city")["amount"].max())
```

```python
# count how many orders per city
print(orders.groupby("city")["order_id"].count())
```

```python
# dataframe.groupby("column")["value_column"].agg["sum", "mean", "count"]
print(orders.groupby("city")["amount"].agg(["sum", "mean", "count"]))
```

```python
# groupby with multiple columns
# dataframe.groupby(["column1", "column2"]).sum()
print(orders.groupby(["city", "products"])["amount"].agg(["sum","count"]))
```

```python
# grouped_result.reset_index()
# .reset_index() -- moves the grouping column (currently the index) back into an ordinary column by replacing the index with a fresh 0, 1, 2....

city_totals = orders.groupby("city")["amount"].sum().reset_index()
print(city_totals)
```

```python
# in case there are diffrent column names then we can use this method
inner_new = orders.merge(customers, left_on="customer_id", right_on="cust_id", how="inner")
print(inner_new[["order_id", "customer_id", "customer_name", "amount"]])
```

```python
left_join = orders.merge(customers, left_on="customer_id", right_on="cust_id", how="left")
print(left_join)
print(left_join.shape)
```

```python
right_join = orders.merge(customers, left_on="customer_id", right_on="cust_id", how="right")
print(right_join)
print(right_join.shape)
```

```python
outer_join = orders.merge(customers, left_on="customer_id", right_on="cust_id", how="outer")
print(outer_join)
print(outer_join.shape)
```

```python
cross_join = orders.merge(customers, how="cross")
print(cross_join)
print(cross_join.shape)
```

```python
# how much revenue does each CUSTOMER SEGMENT generate
left_join= orders.merge(customers, left_on="customer_id", right_on="cust_id", how="left")
segment_summary = left_join.groupby("segment")["amount"].sum().reset_index()
print(segment_summary)
```

```python
print(orders["amount"].sum())
print(segment_summary["amount"].sum())
```

```python
left_join= orders.merge(customers, left_on="customer_id", right_on="cust_id", how="left")
segment_summary = left_join.groupby("segment", dropna = False)["amount"].sum().reset_index()
print(segment_summary)
# we are using dropna = False to avoid NaN value and give it as a different segment
```

```python
# give me sum, mean of amount per city sorted by Max of sum in descending order
city_summary = orders.groupby("city")["amount"].agg(["sum","mean"]).reset_index()
print(city_summary.sort_values(by = "sum", ascending = False))

```
