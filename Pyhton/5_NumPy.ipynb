```python
import numpy as np
score = np.array([78,92,45,23,98,47])
print (score)
print (score.shape) # this gives (6,) which shows its a one dimensional
print (score.dtype) # this gives int64 which means 64bit of intergers

#np.array([...]) build an array directly from a list,
#shape tells you its dimensions - (6,) means one dimension 6 elements.
#dtype tells you the single, uniform type every element shares

# compare this to a python list, which could just as easily hold
# [78 "nine", True] with no complaint at all - Numpy simply won't allow
# that mixing, and the restriction is exactly the source of its speed
```

```python
import time
import numpy as np

big_list = list(range(1000000))
start = time.time()
doubled_list = []
for i in big_list:
  doubled_list.append( 2 * i)
print("List time:", time.time() - start)


big_array = np.array(big_list)
start1 = time.time()
doubled_array = 2 * big_array
print("Array time :", time.time() - start1)
```

```python

score = np.array([78,92,55,88,64])

print(score)
print(score + 5) # adds 5 to each element
print(score * 2) # multiply each element with 2
print(score / 10) # divide each element by 10
print(score // 10) # return the quotient

print(score) # return the same array
```

```python
score = np.array([78,92,55,88,64])

print(score > 75)
```

```python
score = np.array([78,92,55,88,64,34])

print(score[0]) #first element - same as list
print(score[-1]) #last element - same as list
print(score[1:4]) #starting from 1 to 4(4 is not included)
print(score[0:-1:2]) # it will give every second element  [start:stop:step]

#reverse the numpy array
print(score[ : :-1])
```

```python
# boolean masking

scores = np.array([78,92,55,88,64])
mask = scores > 75
print(mask)

# i want to print the numbers greater than 75
print(scores[mask])  #scores[mask] uses that boolean array as the index itself -
# wherever the mask it true, that element is kept, wherever it false that element is droped
# This single line does the entire job of building a list, loop, check a condition
```

```python
temperature  = np.array([28,30,31,27,29,35])
mask = temperature > 29
# temperature above 29
print(temperature[mask])
```

```python
# a small grid of monthly sales for two products
sales = np.array ([
    [100, 150, 200], # product A : Jan, Feb, March
    [300, 250, 400], # product B : Jan, Feb, March
])
monthly_bonus = np.array([10,20,30]) #one bonus value per month, applies to both products

#what exactly is the total sales for both products in each month
total_sales = sales + monthly_bonus
print(total_sales)
print(sales * monthly_bonus)

print(sales.shape)
print(monthly_bonus.shape)


#broadcasting lines them up by their last dimension - 3 columns match 3 values
# and stretches monthly_bonus down both rows, adding it to each one.

print(total_sales.shape)
```

```python
sales = np.array ([
    [100, 150, 200],
    [300, 250, 400],
])
print(sales.sum(axis=0)) # axis 0 collapses the rows together, givng 1 total per columns
print(sales.sum(axis=1)) # axis 1 collapses the columns together, giving 1 total per row
```
