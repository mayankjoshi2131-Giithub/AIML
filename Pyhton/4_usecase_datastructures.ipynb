```python
orders = [
    {"Customer" : "Ananya" , "amount" : 500},  #order1
    {"Customer" : "Rahul" , "amount" : 700},   #order2
    {"Customer" : "Priya" , "amount" : 300},   #order3
    {"Customer" : "Ananya" , "amount" : 100},  #order4
    {"Customer" : "Rahul" , "amount" : 800},   #order5
]
# we need to print name of the customer along with the total order amount
# { "Ananya" : 600, "Priya" : 300, "Rahul" : 1500}

total_orders = {}
# total_orders = {"Ananya" : 500 + 100 = 600 , "Rahul" : 700 + 800 = 1500, "Priya" : 300}
for order in orders:       #{"Customer" : "Rahul" , "amount" : 800}
  customer = order["Customer"] # "Rahul"
  amount = order["amount"]    #800

  if customer in total_orders: #check if ananya is present in total_orders dict
    total_orders[customer] = total_orders[customer] + amount
    #total_orders["Ananya"] = total_orders["Ananya"] that is 500 + 100 = 600
  else:
    total_orders[customer] = amount   # total_order ["Ananya"] = 500
    # total_orders = {"Ananya" : 500, "Rahul" : 700, "Priya" : 300}

print(total_orders)
```

```python
students = [
    {"name" : "Priya", "score" : 78}, #st = 0
    {"name" : "Rahul", "score" : 55}, #st = 1
    {"name" : "Mayank", "score" : 35} #st = 2
]

# Priya:78
# Rahul:55
# Mayank:35

report_lines =[]

for st in students:    #{"name" : "Mayank", "score" : 35}
  name = st["name"]    #Mayank
  score = st["score"]  #35
  line = name + " : " + str(score) # "Mayank" + ":" + "35"
  report_lines.append(line) # ["Priya : 78", "Rahul : 55", "Mayank : 35"]

print(report_lines)
for i in report_lines:
  print(i)
# there is an alternate method of the same
new_report = "\n" .join(report_lines)
# x = " "
#for i in report_lines:
  # x = x + i + "\n" # " " +"Priya : 78" + "\n"
print(new_report)
```
