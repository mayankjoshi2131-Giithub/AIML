```python
#reverse a string

name = "Mayank"
reverse_word = name[::-1]  #start : stop :step  , name = [start:stop:step] , in this when start and stop have no value by default it start from the end of the string
print(reverse_word)
print(len(name))

reverse = ""
for i in range(len(name) - 1, -1, -1):   #range[8,7,6,5,4,3,2,1,0]
  reverse = reverse + name[i]

print(reverse)
```

```python
students = [
    ["A",92,78],
    ["B",55,88],
    ["C",64,70]
]
for i in students:
  name = i[0]
  subject_1 = i[1]
  subject_2 = i[2]
  average = (subject_1 + subject_2)/2
  print(f"Student name {name} has average of {average}")
```

```python
students = {
    "A":{
        "maths" : 92,
        "Science" : 89
    },
    "B":{
        "maths" : 87,
        "Science" : 83
    },
    "C":{
        "maths" : 70,
        "Science" : 81
    },
    "D":{
        "maths" : 99,
        "Science" : 96
    }
}
for name, marks in students.items():  # for key, value in dict_name.items()
  maths_marks = marks["maths"]
  science_marks = marks["Science"]
  average = (maths_marks + science_marks)/2
  print(f"Average of {name} is {marks}")
```
