```python
flagged_max_amount = 3000
transactions = [1200, -300, 4500, -150, 800]

#amount > flagged amount : skip
#amount < 0 : print debited
#else : print Credited

for i in transactions:
  if(i > flagged_max_amount):
    continue
  if(i < 0):
    print(i,"Debited")
  else:
    print(i,"Credited")
```

```python
def rectangle_area(length,breadth):
  area = length * breadth
  return area

rectangle1 = rectangle_area(45,20)  #here def and rectangle1 must be in the straight line otherwise it will show identation error
print(rectangle1)

rectangle2 = rectangle_area(35,66)
print(rectangle2)
```

```python
def score_summary(scores):     #WAF that returns, total score, average score, best score
  total = sum(scores)          #sum is an inbuilt function which is applied on lists
  average = total/len(scores)  #len is also an inbuilt fn return lenght of list
  best = max(scores)

  return total, average, best
scores = [78,92,33,7,56]
total_score, average_score, best_score = score_summary(scores)
print("The total score of team is:",total_score)
print("The average score of team is:",average_score)
print("The best score in the team is:",best_score)


```

```python
def calculate_price(price, tax_rate, discount):
  total = price + price * (tax_rate/100) - discount
  return total

bill = calculate_price(1200, 18, 230)
print(bill)
```
